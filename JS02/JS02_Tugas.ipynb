```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```


1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

```python
df = pd.read_csv('wbc.csv')
df.head(10)
```

```python
df_clean = df.drop(['id', 'Unnamed: 32'], axis=1)

df_clean.to_csv('wbc_cleaned.csv', index=False)

df_clean.head()
```

2.
Lakukan proses encoding pada kolom "diagnosis".



```python
le = LabelEncoder()
df_clean['diagnosis'] = le.fit_transform(df['diagnosis'])
df_clean.head()
```


3. Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

```python
# mengambil semua kolom yang numerik
kolom_numerik = df_clean.select_dtypes(include=['number']).columns.tolist()

# mengeluarkan kolom 'diagnosis' karena ini adalah Target/Label, tidak perlu di-scale
if 'diagnosis' in kolom_numerik:
    kolom_numerik.remove('diagnosis')

# jaga-jaga jika kolom 'Unnamed: 32' atau 'id' masih nyangkut di memori
for col in ['id', 'Unnamed: 32']:
    if col in kolom_numerik:
        kolom_numerik.remove(col)

# proses standardisasi hanya pada fitur numerik
scaler = StandardScaler()
df_clean[kolom_numerik] = scaler.fit_transform(df_clean[kolom_numerik])

df_clean.head()
```
